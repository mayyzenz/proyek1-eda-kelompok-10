### Deskripsi data dan kamus data

Dataset ini berasal dari Pune Smart City Environmental Sensor Data (Kaggle), kumpulan data dari jaringan sensor kualitas udara dan cuaca yang dipasang
oleh pemerintah kota Pune, India, sebagai bagian dari inisiatif Smart City.
Link: https://www.kaggle.com/datasets/akshman/pune-smartcity-test-dataset.
Lisensi: Data files © Original Authors — hak cipta data dipegang oleh penulis/penyedia data asli (Pune Smart City Development Corporation Limited/PCMC
Smart Sarathi). Dataset dalam laporan ini hanya untuk keperluan analisis dan pembelajaran dengan tetap memberikan atribusi kepada pemilik data asli
sesuai ketentuan Kaggle.

Data ini dikumpulkan oleh Pune Smart City Development Corporation Limited (PSCDCL) dan IISC melalui 10 stasiun sensor IoT yang tersebar di berbagai titik strategis kota. Sensor mencatat data secara otomatis dan berkala (interval 15 menit) selama periode 8 April 2019 - 6 September 2019, mencakup parameter polutan udara, kondisi cuaca, kebisingan, dan lokasi geografis setiap stasiun. Dataset ini terdiri dari 103.205 baris dan 28 kolom.
