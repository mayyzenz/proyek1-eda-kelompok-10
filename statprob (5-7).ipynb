### Statistik Deskriptif Variable Numerik

```python
import numpy as np
import pandas as pd

target_cols = [
    "HUMIDITY",
    "HUMADITY",
    "PM2_MAX",
    "NO2_MAX",
    "SO2_MAX",
    "CO2_MAX",
]

selected_cols = [col for col in target_cols if col in df.columns]

print("Kolom numerik yang dianalisis:", selected_cols)

summary_list = []

for col in selected_cols:
  data = df[col].dropna()

  mean_val = data.mean()
  median_val = data.median()
  mode_val = data.mode()[0] if not data.mode().empty else np.nan

  min_val = data.min()
  max_val = data.max()
  range_val = max_val - min_val
  var_val = data.var()
  std_val = data.std()

  q1_val = data.quantile(0.25)
  q3_val = data.quantile(0.75)
  iqr_val = q3_val - q1_val

  summary_list.append({
      "Kolom": col,
      "Rata-rata (Mean)": mean_val,
      "Median": median_val,
      "Modus": mode_val,
      "Minimum": min_val,
      "Maksimum": max_val,
      "Jangkauan (Range)": range_val,
      "Varians": var_val,
      "Simpangan Baku": std_val,
      "Q1": q1_val,
      "Q3": q3_val,
      "IQR": iqr_val,
  })

df_summary = pd.DataFrame(summary_list).set_index("Kolom").T
display(df_summary.round(2))
```

### **Interpretasi Statistik Deskriptif Variabel Numerik**

1. **HUMIDITY (Kelembapan Udara)**
   - **Rata-rata (Mean)** kelembapan udara adalah sebesar **63.13%**, dengan nilai **Median** sebesar **68.20%**.
   - Karena nilai **Mean < Median**, sebaran data kelembapan udara cenderung **miring ke kiri (*left-skewed*)**.
   - Variabilitas data kelembapan cukup bervariasi dengan **Simpangan Baku** sebesar **21.79%** serta rentang nilai dari minimum **10.17%** hingga maksimum **97.36%**.
   - Sebesar 50% data kelembapan udara berada pada rentang tengah antara **Q1 (48.06%)** dan **Q3 (81.10%)** dengan nilai **IQR sebesar 33.04%**.

2. **PM2_MAX (Konsentrasi Maksimum PM2.5)**
   - Nilai **Rata-rata (Mean)** konsentrasi PM2.5 adalah sebesar **45.82 µg/m³**, sedangkan nilai **Median**-nya adalah **38.10 µg/m³**.
   - Karena nilai **Mean > Median**, sebaran data PM2.5 cenderung **miring ke kanan (*right-skewed*)**. Ini menandakan mayoritas pencatatan berada pada tingkat rendah hingga sedang, namun terdapat beberapa lonjakan nilai tinggi (pencilan) hingga mencapai nilai maksimum **215.40 µg/m³**.
   - Sebesar 50% data tengah berada pada rentang antara **Q1 (22.50 µg/m³)** dan **Q3 (61.30 µg/m³)** dengan **IQR sebesar 38.80 µg/m³**.

3. **NO2_MAX, SO2_MAX, dan CO2_MAX (Kadar Maksimum Gas Polutan)**
   - Seluruh data pada kolom `NO2_MAX`, `SO2_MAX`, dan `CO2_MAX` mencatatkan nilai **0.0** di seluruh sampel.
   - Karena tidak ada variasi data (Rata-rata = 0, Median = 0, dan Simpangan Baku = 0), variabel-variabel gas polutan ini mengindikasikan bahwa tidak ada deteksi konsentrasi atau sensor tidak mencatat nilai di atas nol selama periode pengambilan sampel dataset ini.

### Ringkasan Variable Kategorik

```python
import pandas as pd

cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()

if len(cat_cols) > 0:
    kolom_kategorik = cat_cols[0]
    print(f"Kolom kategorik yang ditemukan dan dianalisis: '{kolom_kategorik}'\n")

    frekuensi = df[kolom_kategorik].value_counts()
    persentase = df[kolom_kategorik].value_counts(normalize=True) * 100

    tabel_kategorik = pd.DataFrame(
        {"Frekuensi": frekuensi, "Persentase (%)": persentase.round(2)}
    )

    display(tabel_kategorik)

    modus_kategori = df[kolom_kategorik].mode()[0]
    print(
        f"\nModus (Kategori Terbanyak): '{modus_kategori}' "
        f"dengan frekuensi {frekuensi.iloc[0]} data ({persentase.iloc[0]:.2f}%)"
    )
else:
    print(
        "Tidak ditemukan kolom bertipe teks/kategorik. Berikut daftar seluruh kolom di dataset:"
    )
    print(df.columns.tolist())
```
